```python
# Atividade 2 - Cálculo de Temperatura para Reação Química

def calcular_temperatura_reacao(T_ambiente):
    if T_ambiente < 15:
        return T_ambiente + 10

    elif 15 <= T_ambiente <= 30:
        return T_ambiente

    else:
        return T_ambiente - 5

T_ambiente = float(input("Digite a temperatura ambiente (°C): "))
T_reacao = calcular_temperatura_reacao(T_ambiente)

print(f"A temperatura da reação deve ser ajustada para {T_reacao:.1f}°C.")
```

```python
# falar sobre a importação com as bibliotecas:
 prompt: Um misturador recebe 2 correntes de alimentação:
# A -> 100 kg/h de sikyção com 10% de sal
# B -> 50 kg/h de solução com 30% de sal
# 1. Qual o valor da vazão e da concentração de sal na saida?
# 2. Quais os novos valores quando a cazão da corrente A diminuir em 50% e sua concentração de sal aumentar em 50%
# TAREFA: ENCAPSULAMENTO DA LOGICA OU MODULARIDADE, COM PARAMETROS COMO ENTRDAS E CLAUSULAS PARA TRATAMENTOS DE EXCEÇÕES/ERROS

def calcular_misturador(vazao_a, conc_sal_a, vazao_b, conc_sal_b):
    """
    Calcula a vazão e a concentração de sal na saída de um misturador.

    Args:
        vazao_a (float): Vazão da corrente de alimentação A (kg/h).
        conc_sal_a (float): Concentração de sal na corrente de alimentação A (%).
        vazao_b (float): Vazão da corrente de alimentação B (kg/h).
        conc_sal_b (float): Concentração de sal na corrente de alimentação B (%).

    Returns:
        tuple: Uma tupla contendo:
            - vazao_saida (float): Vazão de saída do misturador (kg/h).
            - conc_sal_saida (float): Concentração de sal na saída (%).
            - erro (str or None): Mensagem de erro, se houver.
    """
    # Validações
    if not all(isinstance(arg, (int, float)) for arg in [vazao_a, conc_sal_a, vazao_b, conc_sal_b]):
        return None, None, "Erro: Todas as entradas devem ser números."
    if vazao_a < 0 or vazao_b < 0:
        return None, None, "Erro: As vazões de alimentação não podem ser negativas."
    if not (0 <= conc_sal_a <= 1) or not (0 <= conc_sal_b <= 1):
         return None, None, "Erro: As concentrações de sal devem estar entre 0 e 1 (ou 0% e 100%)."


    # Balanço de Massa Total
    vazao_saida = vazao_a + vazao_b

    # Tratar caso de vazão total zero para evitar divisão por zero
    if vazao_saida == 0:
        return 0, 0, "A vazão de saída é zero."

    # Balanço de Massa para o Sal
    massa_sal_a = vazao_a * conc_sal_a
    massa_sal_b = vazao_b * conc_sal_b
    massa_sal_saida = massa_sal_a + massa_sal_b

    # Concentração de Sal na Saída
    conc_sal_saida = massa_sal_saida / vazao_saida

    return vazao_saida, conc_sal_saida, None

# Condições Iniciais
vazao_A_inicial = 100  # kg/h
conc_sal_A_inicial = 0.10  # 10%
vazao_B_inicial = 50   # kg/h
conc_sal_B_inicial = 0.30  # 30%

vazao_saida_inicial, conc_sal_saida_inicial, erro_inicial = calcular_misturador(
    vazao_A_inicial, conc_sal_A_inicial, vazao_B_inicial, conc_sal_B_inicial
)

print("Condições Iniciais:")
if erro_inicial:
    print(erro_inicial)
else:
    print(f"1. Vazão de saída: {vazao_saida_inicial:.2f} kg/h")
    print(f"   Concentração de sal na saída: {conc_sal_saida_inicial:.2f} (ou {conc_sal_saida_inicial * 100:.2f}%)")

# Novas condições
vazao_A_nova = vazao_A_inicial * 0.50  # Diminui em 50%
conc_sal_A_nova = conc_sal_A_inicial * 1.50  # Aumenta em 50%
vazao_B_nova = vazao_B_inicial  # Vazão B permanece a mesma
conc_sal_B_nova = conc_sal_B_inicial  # Concentração B permanece a mesma

vazao_saida_nova, conc_sal_saida_nova, erro_nova = calcular_misturador(
    vazao_A_nova, conc_sal_A_nova, vazao_B_nova, conc_sal_B_nova
)

print("\nNovas Condições:")
if erro_nova:
    print(erro_nova)
else:
    print(f"2. Vazão de saída: {vazao_saida_nova:.2f} kg/h")
    print(f"   Concentração de sal na saída: {conc_sal_saida_nova:.2f} (ou {conc_sal_saida_nova * 100:.2f}%)")

# Pensar sobre os modos seletivos utilizando from e import

# Em seguida falar sobre a instalação de bibliotecas

#Numpy; Matplotlib; Pandas; Scipy

#print(np.random.rand(1))
#print(np.random.sample(5))


# descompactação de sequências em um item ou lista
#lista = [1, 2, 3, 4, 5]
#r, *s = lista
#*v, w = lista

# falar sobre os dados primitivos

inteiro = 10
real = 3.14
complexo = 2 + 3j
string = "Olá, mundo!"
booleano = True  # ou False


#append, insert, remove, pop
```
